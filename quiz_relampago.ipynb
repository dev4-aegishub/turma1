```python
import socket
laptop = socket.gethostname()
perg = [("Quanto e 7 x 8?", "56"), ("Capital do Brasil (sem acento)?", "brasilia"),
        ("Funcao que mostra texto na tela?", "print"), ("Quanto e 2 ** 3?", "8")]

print(f"{laptop} Quiz Relampago!"); pontos = 0

for i, (p, resp) in enumerate(perg, 1):
    if input(f"{i}) {p} ").strip().lower() == resp:
        print("Certo!"); pontos += 1
    else:
        print(f"Era: {resp}")

print(f"Voce fez {pontos} de {len(perg)}")
```
