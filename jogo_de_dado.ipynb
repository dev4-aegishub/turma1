```python
# ============================================================
# JOGO: DUELO DE DADOS
# Jogador x Computador
# ============================================================

# Importa a biblioteca random.
# Ela será usada para sortear números aleatórios de 1 até 6.
import random

# Importa a biblioteca socket.
# Ela permite descobrir o nome do computador onde o jogo está rodando.
import socket


# ------------------------------------------------------------
# CONFIGURAÇÃO INICIAL
# ------------------------------------------------------------

# Descobre o nome do laptop/computador.
laptop = socket.gethostname()

# Cria a variável que guarda os pontos do jogador.
# O jogador começa com zero pontos.
voce = 0

# Cria a variável que guarda os pontos do computador.
# O computador também começa com zero pontos.
pc = 0


# ------------------------------------------------------------
# INÍCIO DO JOGO
# ------------------------------------------------------------

# Mostra o nome do computador e o título do jogo.
print(f"[{laptop}] Duelo de Dados - 3 rodadas!")

# Apenas cria uma linha vazia para deixar a tela organizada.
print()


# ------------------------------------------------------------
# RODADAS
# ------------------------------------------------------------

# O comando range(1, 4) gera:
# 1, 2 e 3.
#
# Portanto, o jogo terá exatamente 3 rodadas.
for rodada in range(1, 4):

    # Espera o jogador pressionar ENTER.
    input(f"Rodada {rodada} - pressione ENTER para rolar os dados...")

    # Sorteia o dado do jogador.
    # randint(1, 6) escolhe aleatoriamente um número entre 1 e 6.
    dado_voce = random.randint(1, 6)

    # Sorteia o dado do computador.
    dado_pc = random.randint(1, 6)

    # Mostra os dois resultados.
    print(f"Você tirou: {dado_voce}")
    print(f"PC tirou:   {dado_pc}")


    # --------------------------------------------------------
    # COMPARAÇÃO DOS DADOS
    # --------------------------------------------------------

    # Se o dado do jogador for maior que o dado do computador...
    if dado_voce > dado_pc:

        # Adiciona 1 ponto ao jogador.
        voce += 1

        # Informa quem ganhou a rodada.
        print("Ponto seu!")


    # Caso o dado do computador seja maior...
    elif dado_pc > dado_voce:

        # Adiciona 1 ponto ao computador.
        pc += 1

        # Informa quem ganhou a rodada.
        print("Ponto do PC!")


    # Se nenhum dos dois números for maior,
    # significa que os dados deram o mesmo valor.
    else:

        # Nesse caso ninguém recebe ponto.
        print("Empate!")


    # --------------------------------------------------------
    # PLACAR
    # --------------------------------------------------------

    # Mostra o placar depois de cada rodada.
    print(f"Placar: Você {voce} x {pc} PC")

    # Linha vazia para organizar a saída.
    print()


# ============================================================
# RESULTADO FINAL
# ============================================================

print("===================================")
print("          RESULTADO FINAL")
print("===================================")

# Mostra o placar final.
print(f"Você {voce} x {pc} PC")

# Se o jogador tiver mais pontos que o computador...
if voce > pc:

    # O jogador venceu.
    print("VITÓRIA! Você ganhou!")


# Se o computador tiver mais pontos...
elif pc > voce:

    # O jogador perdeu.
    print("DERROTA! O computador ganhou.")


# Caso os dois tenham a mesma quantidade de pontos...
else:

    # A partida terminou empatada.
    print("EMPATE!")


# ------------------------------------------------------------
# FIM DO PROGRAMA
# ------------------------------------------------------------

print("===================================")
print("Fim do jogo!")
```
