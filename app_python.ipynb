# Meu App AegisHub - em Python
**Login + menu + 3 jogos (Jogo da Velha, Forca, Calculadora), tudo em Python, rodando no Jupyter.**

Rode a celula abaixo com **Shift+Enter** e va respondendo nas caixinhas que aparecem:

1. Digite **1** para Cadastrar (e-mail + senha) e depois **2** para Entrar.
2. No **MENU**, escolha o jogo: **1** Velha, **2** Forca, **3** Calculadora, **0** Sair.


```python
# ===== Meu App AegisHub - Python puro no Jupyter =====
# Login + menu + 3 jogos (Velha, Forca, Calculadora). Roda com input()/print().
import random

usuarios = {}  # guarda e-mail -> senha (enquanto o notebook estiver aberto)

# ---------- LOGIN ----------
def cadastrar():
    email = input("Novo e-mail: ").strip().lower()
    if email in usuarios:
        print("Ja existe cadastro para esse e-mail.\n"); return
    senha = input("Nova senha: ")
    usuarios[email] = senha
    print("Cadastro criado! Agora faca login.\n")

def entrar():
    email = input("E-mail: ").strip().lower()
    senha = input("Senha: ")
    if usuarios.get(email) == senha:
        print("Bem-vindo, " + email + "!\n"); return email
    print("E-mail ou senha invalidos.\n"); return None

# ---------- JOGO DA VELHA ----------
def jogo_da_velha():
    tab = [" "] * 9
    vencedoras = [(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]
    def mostra():
        print()
        for i in range(0, 9, 3):
            linha = " | ".join(tab[i+j] if tab[i+j] != " " else str(i+j+1) for j in range(3))
            print(" " + linha)
            if i < 6:
                print("---+---+---")
        print()
    vez = "X"
    for _ in range(9):
        mostra()
        escolha = input("Jogador " + vez + ", escolha a casa (1-9): ").strip()
        if not escolha.isdigit() or not (1 <= int(escolha) <= 9):
            print("Digite um numero de 1 a 9."); continue
        p = int(escolha) - 1
        if tab[p] != " ":
            print("Essa casa ja esta ocupada, tente outra."); continue
        tab[p] = vez
        if any(tab[a] == tab[b] == tab[c] == vez for a, b, c in vencedoras):
            mostra(); print("Jogador " + vez + " venceu!\n"); return
        vez = "O" if vez == "X" else "X"
    mostra(); print("Deu velha! (empate)\n")

# ---------- JOGO DA FORCA ----------
def jogo_da_forca():
    palavras = ["python", "jupyter", "codigo", "aegishub", "notebook", "programar", "futuro"]
    palavra = random.choice(palavras)
    acertos = set(); erros = 0
    while erros < 6:
        mostra = " ".join(c if c in acertos else "_" for c in palavra)
        print("\nPalavra: " + mostra + "   (erros: " + str(erros) + "/6)")
        if all(c in acertos for c in palavra):
            print("PARABENS! Voce acertou: " + palavra + "\n"); return
        letra = input("Uma letra: ").strip().lower()
        if not letra:
            continue
        letra = letra[0]
        if letra in palavra:
            acertos.add(letra); print("Boa!")
        else:
            erros += 1; print("Nao tem essa letra.")
    print("\nFim! A palavra era: " + palavra + "\n")

# ---------- CALCULADORA ----------
def calculadora():
    try:
        a = float(input("Primeiro numero: "))
        op = input("Operacao (+ - * /): ").strip()
        b = float(input("Segundo numero: "))
    except ValueError:
        print("Numero invalido.\n"); return
    if op == "+": r = a + b
    elif op == "-": r = a - b
    elif op == "*": r = a * b
    elif op == "/": r = "Erro: divisao por zero" if b == 0 else a / b
    else:
        print("Operacao invalida.\n"); return
    print("Resultado: " + str(a) + " " + op + " " + str(b) + " = " + str(r) + "\n")

# ---------- PROGRAMA PRINCIPAL ----------
print("=== Meu App AegisHub (Python) ===\n")
usuario = None
while usuario is None:
    escolha = input("Digite 1 para Cadastrar ou 2 para Entrar: ").strip()
    if escolha == "1":
        cadastrar()
    elif escolha == "2":
        usuario = entrar()
    else:
        print("Opcao invalida.")

while True:
    print("----- MENU -----")
    print("1 - Jogo da Velha")
    print("2 - Jogo da Forca")
    print("3 - Calculadora")
    print("0 - Sair")
    op = input("Escolha uma opcao: ").strip()
    if op == "1":
        jogo_da_velha()
    elif op == "2":
        jogo_da_forca()
    elif op == "3":
        calculadora()
    elif op == "0":
        print("Ate mais, " + usuario + "!")
        break
    else:
        print("Opcao invalida.\n")
```
