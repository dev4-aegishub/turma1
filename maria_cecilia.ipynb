```python
É um jogo para tentar acertar o número 
```

```python
import random, socket
laptop = socket.gethostname()
segredo = random.randint(1, 20)
print(f"[{laptop}] Adivinhe o numero de 1 a 20! Voce tem 5tentativas.")
for tentativa in range(1, 6):
    palpite = int(input(f"Tentativa {tentativa}/5 - seu numero: "))
    if palpite == segredo:
        print("ACERTOU! Parabens!")
        break
    elif palpite < segredo:
        print("O numero e MAIOR.")
    else:
        print("O numero e MENOR.")
    
else:
    print(f"Acabaram as tentativas. O numero era {segredo}.")
```
